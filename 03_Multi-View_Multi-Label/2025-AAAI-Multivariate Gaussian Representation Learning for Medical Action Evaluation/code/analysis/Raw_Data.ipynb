```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"./Data/CPREval_Annotation.csv")

selected_columns = ['Additional Label', 'Main Label', 'annotation_id', 'annotator', 'id']
filtered_df = df[selected_columns]

main_labels = filtered_df['Main Label'].unique()

num_rows = (len(main_labels) + 1) // 2
num_cols = 2
fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, 5 * num_rows))
fig.suptitle('Additional Label Distribution for Each Main Label', fontsize=16)

if len(main_labels) == 1:
    axes = [axes]

for i, main_label in enumerate(main_labels):
    ax = axes[i // num_cols, i % num_cols] if num_rows > 1 else axes[i % num_cols]
    subset = filtered_df[filtered_df['Main Label'] == main_label]
    additional_label_counts = subset['Additional Label'].value_counts()
    additional_label_counts.plot(kind='bar', ax=ax, color='skyblue')
    ax.set_title(f'Main Label: {main_label}')
    ax.set_xlabel('Additional Label')
    ax.set_ylabel('Count')
    ax.tick_params(axis='x', rotation=45)

for i in range(len(main_labels), num_rows * num_cols):
    fig.delaxes(axes.flatten()[i])

plt.subplots_adjust(hspace=0.5, wspace=0.3)

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
selected_wrong_columns = [
    'Main Label', 
    'Additional Label 1', 
    'Additional Label 2', 
    'Additional Label 3', 
    'Additional Label 4', 
    'Additional Label 5'
]
wrong_df = filtered_df_sta[selected_wrong_columns]
wrong_df = wrong_df[wrong_df['Main Label'] != 'Correct']
wrong_df = wrong_df.replace({None: pd.NA})
print(wrong_df.head(20))
```

```python
error_columns = [
    'Hand-Double Hand-Overlap', 'Hand-Double Hand-Cross',
    'Hand-Double Hand-Two Fist', 'Hand-One Hand-Hand', 'Hand-One Hand-Fist',
    'Angle-Arm Bend', 'Angle-Front Skew', 'Angle-Back Skew', 'Angle-Left Skew',
    'Angle-Right Skew', 'Body-Waist Unstable', 'Position-Wrong Position',
    'Position-Movement', 'Body-Thigh Antevert', 'Freq-Excessive Momentum',
    'Freq-Low Frequency', 'Freq-High Frequency', 'Depth-Insufficient Press',
    'Body-Thigh Tilt Back', 'Depth-Insufficient Rebound', 'Depth-Excess'
]

melted_df = wrong_df.melt(
    value_vars=selected_wrong_columns,
    value_name='Error',
    ignore_index=False
).dropna(subset=['Error'])

dummies = pd.get_dummies(melted_df['Error'], dtype=int)
dummies = dummies.groupby(level=0).max()

dummies = dummies.reindex(columns=error_columns, fill_value=0)
final_df = wrong_df[['Main Label']].join(dummies)

# 查看结果
print(final_df.head())
final_df.to_csv(r'./Data/final_annotation.csv', index=False, encoding='utf-8') 
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

error_df = final_df[error_columns]

nonzero_var_cols = error_df.columns[error_df.var() > 0].tolist()
error_df_filtered = error_df[nonzero_var_cols]

corr_matrix = error_df_filtered.corr()

plt.figure(figsize=(15, 13))

heatmap = sns.heatmap(
    corr_matrix, 
    annot=True, 
    fmt=".2f", 
    cmap='coolwarm', 
    cbar=True, 
    square=True, 
    linewidths=0.5, 
    linecolor='white',
    annot_kws={'size': 8}
)

heatmap.set_xticklabels(
    heatmap.get_xticklabels(), 
    rotation=45, 
    ha='right', 
    fontsize=10
)
heatmap.set_yticklabels(
    heatmap.get_yticklabels(), 
    rotation=0, 
    fontsize=10
)

plt.title('Error type correlation heatmap', fontsize=14, pad=20)

plt.tight_layout()
plt.savefig(
    r'./Data/error_correlation_heatmap.png', 
    dpi=300, 
    bbox_inches='tight'
)

plt.show()
```

```python
from sklearn.feature_selection import mutual_info_regression
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

def mutual_information_matrix(df):
    n = df.shape[1]
    mi_matrix = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            mi_matrix[i, j] = mutual_info_regression(
                df.iloc[:, [i]], df.iloc[:, j], discrete_features=False
            )[0]
    return mi_matrix

error_df = final_df[error_columns]

nonzero_var_cols = error_df.columns[error_df.var() > 0].tolist()
error_df_filtered = error_df[nonzero_var_cols]

mi_matrix = mutual_information_matrix(error_df_filtered)

mi_df = pd.DataFrame(mi_matrix, index=nonzero_var_cols, columns=nonzero_var_cols)

plt.figure(figsize=(15, 13))

heatmap = sns.heatmap(
    mi_df, 
    annot=True, 
    fmt=".2f", 
    cmap='coolwarm', 
    cbar=True, 
    square=True, 
    linewidths=0.5, 
    linecolor='white',
    annot_kws={'size': 8}
)

heatmap.set_xticklabels(
    heatmap.get_xticklabels(), 
    rotation=45, 
    ha='right', 
    fontsize=10
)
heatmap.set_yticklabels(
    heatmap.get_yticklabels(), 
    rotation=0, 
    fontsize=10
)

plt.title('Error Type Non-linear Correlation (Mutual Information) Heatmap', fontsize=14, pad=20)

plt.tight_layout()
plt.savefig(
    r'./Data/error_mutual_information_heatmap.png', 
    dpi=300, 
    bbox_inches='tight'
)

plt.show()
```
